```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# sentences

sent = [
    "The quick brown fox jumps over the lazy dog",
    "I love programming in Python",
    "The sky is blue and beautiful",
    "I like to eat pizza",
    "I love to play cricket",
    "I love to watch movies",
]
```

```python
sent
```

```python
# define the vocabulary size
vocab_size = 10000

# one hot encode the sentences
encoded_sent = [one_hot(sent, vocab_size) for sent in sent]
encoded_sent

```

```python
# word embedding representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np


sent_length = 8

embedded_docs = pad_sequences(encoded_sent, padding='pre', maxlen=sent_length)

```

```python
embedded_docs
```

```python
# feature representation

dim = 10

model = Sequential()
model.add(Embedding(vocab_size, dim, input_length=sent_length))
model.compile('adam', 'mse')
model.summary()
```

```python
model.predict(embedded_docs)
```
